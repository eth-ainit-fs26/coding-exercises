### 1.4 Dictionaries
Dictionaries store key-value pairs, throughout the course you you'll see dictionaries everywhere for tracking state

````python
# Lets define a dictonay similar to the one you will encounter
current_state = {
    "chosen": [-7, 5],
    "remaining": [3, -1, -2, 7, -5],
    "rejected": set(),
    "current_sum": -2,
    "step": 3
}
````

This dictionary has 4 keys, each pointing to a different type of value:

| Key          | Value Type | What It Represents                              |
|--------------|-----------|--------------------------------------------------|
| chosen       | List      | Numbers the agents have already selected         |
| remaining    | List      | Numbers still available to choose from           |
| rejected     | Set       | Numbers the agents have rejected (empty for now) |
| current_sum  | Integer   | Running total of chosen numbers                  |

This dictonary allows you to look up a key (the numbers 7, 5, 3, -1, -2, -5) and sum the value of the selected keys

````python
# Accessing values
print(f"Chosen numbers: {current_state['chosen']}")
print(f"Current sum: {current_state['current_sum']}")
````

````python
# Updating state
current_state["current_sum"] = 0
current_state["chosen"].append(2)
print(f"Chosen numbers: {current_state}")
````

### 1.5 Sets
Sets store unique items

````python
# Creating sets
unique_numbers = {1, 2, 3, 3, 4, 4, 5}
# duplicates are automatically removed
print(f"Set with duplicates removed: {unique_numbers}")
````

````python
# Adding items
unique_numbers.add(6)
print(f"After adding: {unique_numbers}")
````

````python
# Checking membership
print(f"Is 3 in set? {3 in unique_numbers}")
print(f"Is 10 in set? {10 in unique_numbers}")
````

````python
# Set operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print(f"Union: {set_a | set_b}")  # all items from both
print(f"Intersection: {set_a & set_b}")  # items in both
````

## **Part 2: Control Flow**

### 2.1 If Statements
An if statement runs a block of code if the condition is true

In this example we condition the variable age to determine if they are an adult or a minor

````python
age = 18
````

````python
if age >= 18:
    print("You are an adult")
else:
    print("You are a minor")
````

Let's add another variable, student, that will allow us to restrain or code even more

````python
# 0 meaning not a student and 1 is a student
student = 0
````

````python
if age >= 18 and student == 1:
    print("You are an adult and a student")
elif age>= 18 and student == 0:
    print("You are an adult")
elif age<= 18 and student == 1:
    print("You are a minor and a student")
else:
  print("You are a minor")
````

There are many operators that can help you condition and compare statements:

| Operator | Name | What It Does |

| == | Equal to | Checks if two values are equal |
| != | Not equal to | Checks if two values are NOT equal |
| > | Greater than | Checks if left value is greater than right |
| < | Less than | Checks if left value is less than right |
| >= | Greater than or equal to | Checks if left is greater than or equal to right |
| <= | Less than or equal to | Checks if left is less than or equal to right |
| is | Identity | Checks if two objects are the same object |

### 2.1 functions
In this course you will often see functions and at times be instructed to call a certain funtion. A function bundles a set of instructions that you want to use repeatedly. It has three important features you will need to focus on:

*   Function name
*   Arguments
*   Output

````python
# Defining a simple function
def greet(name):
    """This function greets a person"""
    return f"Hello, {name}!"
````

Here the function name is **greet**

The argument is **name** Here you can add a string or any variable, as it is not restricted to a type

The output is always followed by a return in this function it is a string

````python
# Calling the function
message = greet("Alice")
print(message)
````

````python
# Function with multiple parameters
def calculate_fare(base_price, class_multiplier):
  """base_price must be float
     Cclass mmultiplier should be 1 or 2"""
    return base_price * class_multiplier
````

````python
# Calling the function
fare = calculate_fare(base_price = 10.50, class_multiplier = 2)
print(fare)
````

````python
# Function with default parameter
def passenger_info(name, age, survived=False):
    return {
        "name": name,
        "age": age,
        "survived": survived
    }
````

# NumPy Statistical Functions - Quick Reference

| Function | What It Does |
|----------|--------------|
| `np.mean()` | Calculates the average (sum of all values divided by count) |
| `np.median()` | Finds the middle value when data is sorted |
| `np.std()` | Measures how spread out the data is (standard deviation) |
| `np.sum()` | Adds up all values in the array |
| `np.min()` | Finds the smallest value in the array |
| `np.max()` | Finds the largest value in the array |
| `np.var()` | Measures spread (average of squared deviations from mean) |
| `np.exp()` | Calculates exponential (e raised to each element) |
| `np.log()` | Calculates natural logarithm of each element |
| `np.abs()` | Returns absolute value of each element |


````python

survival = np.array([1, 0, 1, 0, 1, 1, 0, 0, 1, 0])

print(f"Survival data: {survival}")
print(f"Mean (survival rate): {np.mean(survival):.2%}")
print(f"Median: {np.median(survival)}")
print(f"Standard deviation: {np.std(survival):.3f}")
print(f"Sum of survivors: {np.sum(survival)}")
print(f"Minimum: {np.min(survival)}")
print(f"Maximum: {np.max(survival)}")
````
